```python
from src.state.replanning_state import ReplanningState
from src.agents.replanning.planner import Planner
from src.agents.replanning.executor import Executor
from src.agents.replanning.replanning import Replanner
```

```python
goal = "Find current BCCI president"

state = ReplanningState(goal)

planner = Planner()
executor = Executor()
replanner = Replanner()

state = planner.create_plan(state)

while True:

    state = executor.execute(state)

    print("=" * 50)
    print("Completed:", state.completed_tasks)
    print("Failed:", state.failed_tasks)
    print("Pending:", state.pending_tasks)
    print("Replan Count:", state.replan_count)
    print("Failed Tasks:", len(state.failed_tasks))
    print("=" * 50)

    if not state.failed_tasks and not state.pending_tasks:
        print("SUCCESS")
        break

    if state.replan_count >= state.max_replans:
        print("MAX REPLANS REACHED")
        break

    if state.failed_tasks:
        state = replanner.replan(state)
```
